## Creating User IDs
We assume every the data is arranged in an ordered format, so like first 10 images belong to first person, next 10 to the next and so on

```python
import numpy as np

# User IDs: 0,0,...,0,1,1,...,99
user_ids = np.repeat(np.arange(100), 10)

# Sample numbers: 0-9 repeated for each user
sample_ids = np.tile(np.arange(10), 100)

# Add them to the dataframe
df["user_id"] = user_ids
df["sample_id"] = sample_ids
```

## Dividing into Test & Train Datasets

```python
train_df = df[df["sample_id"] < 5].reset_index(drop=True)
test_df  = df[df["sample_id"] >= 5].reset_index(drop=True)
```

```python
print(train_df.shape)
print(test_df.shape)
```

## Calculating Euclidean Distance and Cosine Similarities

```python
import numpy as np
from sklearn.metrics.pairwise import euclidean_distances, cosine_similarity

# Feature vectors only
X_train = train_df.iloc[:, :144].to_numpy(dtype=float)
X_test = test_df.iloc[:, :144].to_numpy(dtype=float)

# Corresponding user IDs
train_users = train_df["user_id"].to_numpy()
test_users = test_df["user_id"].to_numpy()
```

```python
# Euclidean distance matrix
euclidean_matrix = euclidean_distances(X_test, X_train)

# Cosine similarity matrix
cosine_matrix = cosine_similarity(X_test, X_train)
```

```python
print(euclidean_matrix.shape)
print(cosine_matrix.shape)

print(euclidean_matrix[:5, :5])
print(cosine_matrix[:5, :5])
```

## Getting the Genuine and Impostor Distributions

```python
# Lists to store scores
genuine_euclidean = []
impostor_euclidean = []

genuine_cosine = []
impostor_cosine = []

# Separate the scores
for i in range(len(test_users)):
    for j in range(len(train_users)):
        if test_users[i] == train_users[j]:
            genuine_euclidean.append(euclidean_matrix[i, j])
            genuine_cosine.append(cosine_matrix[i, j])
        else:
            impostor_euclidean.append(euclidean_matrix[i, j])
            impostor_cosine.append(cosine_matrix[i, j])
```

```python
print("Genuine Euclidean:", len(genuine_euclidean))
print("Impostor Euclidean:", len(impostor_euclidean))

print("Genuine Cosine:", len(genuine_cosine))
print("Impostor Cosine:", len(impostor_cosine))
```

## Calculating FAR & FRR

### Euclidean FAR & FRR
Calculating FAR & FRR for Euclidean distance with multiple different thresholds.

```python
import numpy as np

euclidean_thresholds = np.linspace(
    min(genuine_euclidean + impostor_euclidean),
    max(genuine_euclidean + impostor_euclidean),
    200
)
```

```python
far_euclidean = []
frr_euclidean = []

genuine = np.array(genuine_euclidean)
impostor = np.array(impostor_euclidean)

for threshold in euclidean_thresholds:

    # Genuine rejected
    FRR = np.mean(genuine > threshold)

    # Impostor accepted
    FAR = np.mean(impostor <= threshold)

    far_euclidean.append(FAR)
    frr_euclidean.append(FRR)
```

```python
far_frr_df = pd.DataFrame({
    "Threshold": np.concatenate([euclidean_thresholds, euclidean_thresholds]),
    "Rate": np.concatenate([far_euclidean, frr_euclidean]),
    "Metric": ["FAR"] * len(euclidean_thresholds) +
              ["FRR"] * len(euclidean_thresholds)
})

fig = px.line(
    far_frr_df,
    x="Threshold",
    y="Rate",
    color="Metric",
    title="FAR and FRR vs Threshold (Euclidean Distance)"
)

fig.update_layout(
    xaxis_title="Euclidean Distance Threshold",
    yaxis_title="Error Rate"
)

fig.show()
```

### Cosine FAR & FRR

```python
cosine_thresholds = np.linspace(
    min(genuine_cosine + impostor_cosine),
    max(genuine_cosine + impostor_cosine),
    200
)
```

```python
far_cosine = []
frr_cosine = []

genuine = np.array(genuine_cosine)
impostor = np.array(impostor_cosine)

for threshold in cosine_thresholds:

    # Genuine rejected
    FRR = np.mean(genuine < threshold)

    # Impostor accepted
    FAR = np.mean(impostor >= threshold)

    far_cosine.append(FAR)
    frr_cosine.append(FRR)
```

```python
far_frr_df = pd.DataFrame({
    "Threshold": np.concatenate([cosine_thresholds, cosine_thresholds]),
    "Rate": np.concatenate([far_cosine, frr_cosine]),
    "Metric": ["FAR"] * len(cosine_thresholds) +
              ["FRR"] * len(cosine_thresholds)
})

fig = px.line(
    far_frr_df,
    x="Threshold",
    y="Rate",
    color="Metric",
    title="FAR and FRR vs Threshold (Cosine Similarity)"
)

fig.update_layout(
    xaxis_title="Cosine Similarity Threshold",
    yaxis_title="Error Rate"
)

fig.show()
```

## Receiver Originality Characteristics
Comparing ROC with both euclidean distance and cosine similarity

```python
roc_compare = pd.DataFrame({
    "FAR": np.concatenate([far_euclidean, far_cosine]),
    "GAR": np.concatenate([
        1 - np.array(frr_euclidean),
        1 - np.array(frr_cosine)
    ]),
    "Method": (["Euclidean"] * len(far_euclidean)) +
              (["Cosine"] * len(far_cosine))
})

fig = px.line(
    roc_compare,
    x="FAR",
    y="GAR",
    color="Method",
    title="ROC Curve Comparison"
)

fig.update_layout(
    xaxis_title="False Acceptance Rate (FAR)",
    yaxis_title="Genuine Acceptance Rate (GAR)"
)

fig.show()
```

## Equal Error Rate
The point where FAR = FRR. The lower the EER, the better the system.

### Euclidean EER

```python
# Convert to NumPy arrays
far_e = np.array(far_euclidean)
frr_e = np.array(frr_euclidean)

# Index where FAR and FRR are closest
eer_idx_e = np.argmin(np.abs(far_e - frr_e))

eer_threshold_e = euclidean_thresholds[eer_idx_e]
eer_e = (far_e[eer_idx_e] + frr_e[eer_idx_e]) / 2

print("=== Euclidean ===")
print(f"EER Threshold : {eer_threshold_e:.3f}")
print(f"EER           : {eer_e:.4f}")
```

### Cosine EER

```python
far_c = np.array(far_cosine)
frr_c = np.array(frr_cosine)

eer_idx_c = np.argmin(np.abs(far_c - frr_c))

eer_threshold_c = cosine_thresholds[eer_idx_c]
eer_c = (far_c[eer_idx_c] + frr_c[eer_idx_c]) / 2

print("=== Cosine ===")
print(f"EER Threshold : {eer_threshold_c:.5f}")
print(f"EER           : {eer_c:.4f}")
```

Using Cosine similarity gives us an EER of 0.1146 compared to 0.1558 with Euclidean Distance. Therefore cosine similarity is an objectively superior method for a biometric identification system.

## Decidability Index
The decidability Index tells us how "decidable" the data is, by comparing the means and standard deviations of the genuine and impostor scores. The higher the decidability the better the system

### Euclidean Decidability

```python
mu_g = np.mean(genuine_euclidean)
mu_i = np.mean(impostor_euclidean)

sigma_g = np.std(genuine_euclidean)
sigma_i = np.std(impostor_euclidean)

d_euclidean = abs(mu_g - mu_i) / np.sqrt((sigma_g**2 + sigma_i**2) / 2)

print("Euclidean Decidability Index:", d_euclidean)
```

### Cosine Decidability

```python
mu_g = np.mean(genuine_cosine)
mu_i = np.mean(impostor_cosine)

sigma_g = np.std(genuine_cosine)
sigma_i = np.std(impostor_cosine)

d_cosine = abs(mu_g - mu_i) / np.sqrt((sigma_g**2 + sigma_i**2) / 2)

print("Cosine Decidability Index:", d_cosine)
```

As expected, Cosine Similarity has a higher decidability index.

Decidability Index is inversely proportional to Equal Error Rate
